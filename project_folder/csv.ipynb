```python
import mysql.connector
import csv

# Connect to the database
db = mysql.connector.connect(
    host="localhost",
    user="root",
    password="",
    database="registrat"
)

cursor = db.cursor()
cursor.execute("SELECT * FROM user_entries")
rows = cursor.fetchall()

# Save data into CSV
with open('user_entries.csv', mode='w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    # Optional: write the header
    writer.writerow(['id', 'name', 'committee', 'sessions_attended', 'entry_time'])
    writer.writerows(rows)

print("Data exported successfully to user_entries.csv")
```
